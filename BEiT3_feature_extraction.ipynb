```python
import os
import numpy as np
import pandas as pd
import glob
import torch
import cv2
from PIL import Image
from tqdm import tqdm
import modeling_finetune
from torchvision import transforms
from torchvision.datasets.folder import default_loader
from transformers import XLMRobertaTokenizer
from timm import create_model
from timm.data.constants import IMAGENET_DEFAULT_MEAN, IMAGENET_DEFAULT_STD, IMAGENET_INCEPTION_MEAN, IMAGENET_INCEPTION_STD
from IPython.display import clear_output
import torch_xla.core.xla_model as xm
```

```python
keyframes_dir = "/kaggle/input/new-index-final/keyframes"
output_dir = "/kaggle/working/BEIT3-features"

if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
device = xm.xla_device()
print(f"Using device: {device}")
checkpoint = torch.load('/kaggle/working/unilm/beit3/beit3_large_patch16_384_coco_retrieval.pth')
model = create_model('beit3_large_patch16_384_retrieval')
model.load_state_dict(checkpoint['model'])
model.eval().to(device)
clear_output()
```

```python
transform = transforms.Compose([
            transforms.Resize((384, 384), interpolation=3),
            transforms.ToTensor(),
            transforms.Normalize(mean=IMAGENET_INCEPTION_MEAN, std=IMAGENET_INCEPTION_STD)
        ])
loader = default_loader
```

```python
def load_keyframes(video_id):
    """
    Load keyframes from the directory corresponding to the video_id.
    """
    frames_dir = os.path.join(keyframes_dir, video_id)
    keyframes = []
    for frame_file in sorted(os.listdir(frames_dir)):
        if frame_file.endswith('.jpg'):
            frame_path = os.path.join(frames_dir, frame_file)
            frame = cv2.imread(frame_path)
            keyframes.append(frame)
    return keyframes

def preprocess_frames(frames):
    """
    Preprocess the frames for the model.
    """
    preprocessed_frames = []
    for frame in frames:
        image = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        preprocessed_image = transform(image)
        '''
        image = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        preprocessed_image = vis_processors["eval"](image).unsqueeze(0).to(device)'''
        preprocessed_frames.append(preprocessed_image)
    preprocessed_frames = torch.stack(preprocessed_frames).to(device)
    return preprocessed_frames

def process_frames_in_batches(frames, batch_size=16):
    """
    Process frames in batches and extract features.
    """
    features = []
    for i in range(0, len(frames), batch_size):
        #batch_frames = frames[i:i+batch_size]
        #batch_frames_tensor = torch.cat(batch_frames).half()
        batch_frames_tensor = frames[i:i+batch_size]
        with torch.no_grad():
            '''
            sample = {"image": batch_frames_tensor, "text_input": [""]}
            batch_features = model.extract_features(sample, mode="image")
            batch_features = batch_features.image_embeds_proj.cpu().numpy()
            batch_features = batch_features.reshape(batch_features.shape[0], -1)
            features.append(batch_features)'''
            batch_features, _ = model(image=batch_frames_tensor, only_infer=True) #Error processing video L01_V016: 'tuple' object has no attribute 'cpu'
            batch_features = batch_features.cpu().numpy()
            #print(batch_features.shape)
            features.append(batch_features)
        del batch_frames_tensor
        xm.mark_step()
    return np.concatenate(features, axis=0)
```

```python
video_dirs = [f for f in os.listdir(keyframes_dir) if (f.startswith('L25_V044') or f.startswith('L25_V054') or f.startswith('L25_V057') or f.startswith('L25_V064') or f.startswith('L25_V065') or f.startswith('L25_V071')) and os.path.isdir(os.path.join(keyframes_dir, f))]
for video_dir in video_dirs:
    video_id = video_dir
    print(f"Processing keyframes for video: {video_id}")
    try:
        frames = load_keyframes(video_id)
        preprocessed_frames = preprocess_frames(frames)
        features = process_frames_in_batches(preprocessed_frames)
        output_path = os.path.join(output_dir, f"{video_id}.npy")
        np.save(output_path, features)
        print(f"Saved features for {video_id} to {output_path}")
    except Exception as e:
        print(f"Error processing video {video_id}: {e}")
```
