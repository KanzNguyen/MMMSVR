```python
import os
import torch
import numpy as np
from PIL import Image
import cv2
import torch_xla.core.xla_model as xm
from lavis.models import load_model_and_preprocess
```

```python
video_dir = "/kaggle/input/aic2024-round1-data/video"
output_dir = "/kaggle/working/BLIP2-features"
prefix = "L14"

if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
device = xm.xla_device()
print(f"Using device: {device}")
model, vis_processors, txt_processors = load_model_and_preprocess(
    name="blip2_feature_extractor",
    model_type="pretrain",
    is_eval=True,
    device=device
)
```

```python
def extract_frames(video_path, frame_rate=2):
    video_capture = cv2.VideoCapture(video_path)
    frames = []
    fps = video_capture.get(cv2.CAP_PROP_FPS)
    interval = int(fps / frame_rate)
    success, frame = video_capture.read()
    count = 0
    while success:
        if count % interval == 0:
            frames.append(frame)
        success, frame = video_capture.read()
        count += 1
    video_capture.release()
    return frames

def preprocess_frames(frames):
    preprocessed_frames = []
    for frame in frames:
        image = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        preprocessed_image = vis_processors["eval"](image).unsqueeze(0).to(device)
        preprocessed_frames.append(preprocessed_image)
    return preprocessed_frames

def process_frames_in_batches(frames, batch_size=16):
    features = []
    for i in range(0, len(frames), batch_size):
        batch_frames = frames[i:i+batch_size]
        batch_frames_tensor = torch.cat(batch_frames).half()
        with torch.no_grad():
            sample = {"image": batch_frames_tensor, "text_input": [""]}
            batch_features = model.extract_features(sample, mode="image")
            batch_features = batch_features.image_embeds_proj.cpu().numpy()
            batch_features = batch_features.reshape(batch_features.shape[0], -1)
            #print(batch_features.shape)
            features.append(batch_features)
        del batch_frames_tensor
        xm.mark_step()
    return np.concatenate(features, axis=0)
```

```python
# Get all files with the specified prefix
video_files = [f for f in os.listdir(video_dir) if f.startswith(prefix) and f.endswith('.mp4')]
for video_file in video_files:
    video_path = os.path.join(video_dir, video_file)
    video_id = os.path.splitext(video_file)[0]
    print(f"Processing video: {video_id}")
    try:
        frames = extract_frames(video_path)
        preprocessed_frames = preprocess_frames(frames)
        features = process_frames_in_batches(preprocessed_frames)
        output_path = os.path.join(output_dir, f"{video_id}.npy")
        np.save(output_path, features)
        print(f"Saved features for {video_id} to {output_path}")
    except Exception as e:
        print(f"Error processing video {video_id}: {e}")
```
