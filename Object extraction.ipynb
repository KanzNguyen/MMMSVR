```python
import cv2
import os
import numpy as np
import csv
import torch
from ultralytics import YOLO
import warnings
from tqdm import tqdm
import matplotlib.pyplot as plt
warnings.filterwarnings("ignore", category=FutureWarning, module="sklearn")
```

```python
model = YOLO('yolov10x.pt')  

input_dir = "/kaggle/input/new-index-final/keyframes"
output_base_dir = "/kaggle/working/"

def process_video_folder(video_folder_path, video_name):
    output_dir = os.path.join(output_base_dir, video_name)
    os.makedirs(output_dir, exist_ok=True)

    for img_name in sorted(os.listdir(video_folder_path)):
        if img_name.endswith(".jpg"): 
            img_path = os.path.join(video_folder_path, img_name)

            results = model(img_path)

            image_base_name = os.path.splitext(img_name)[0]

            for i, result in enumerate(results[0].boxes.xyxy):
                x1, y1, x2, y2 = map(int, result)

                img = cv2.imread(img_path)

                cropped_img = img[y1:y2, x1:x2]

                cropped_img_name = f"{video_name}_{image_base_name}_{i:03d}.jpg"
                cropped_img_path = os.path.join(output_dir, cropped_img_name)

                cv2.imwrite(cropped_img_path, cropped_img)

for video_folder in sorted(os.listdir(input_dir)):
    if video_folder.startswith("L28") or video_folder.startswith("L29") or video_folder.startswith("L30"):
      video_folder_path = os.path.join(input_dir, video_folder)

      if os.path.isdir(video_folder_path):  
        print(f"Processing video folder: {video_folder}")
        process_video_folder(video_folder_path, video_folder)
```
